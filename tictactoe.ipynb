```python
import pandas as pd
import matplotlib.pyplot as plt

# Question
print("Q: Implement Tic-Tac-Toe AI using Rule-Based Logic.")

# Board (0-8 positions)
board = [' ']*9

# Display Board
def print_board():
    for i in range(0,9,3):
        print(board[i:i+3])

# Check Win
def check_win(player):
    wins = [(0,1,2),(3,4,5),(6,7,8),
            (0,3,6),(1,4,7),(2,5,8),
            (0,4,8),(2,4,6)]
    return any(board[a]==board[b]==board[c]==player for a,b,c in wins)

# Rule-Based AI Move
def ai_move():
    # Rule 1: Win if possible
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            if check_win('O'):
                return i
            board[i] = ' '
    
    # Rule 2: Block opponent
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'X'
            if check_win('X'):
                board[i] = 'O'
                return i
            board[i] = ' '
    
    # Rule 3: Take center
    if board[4] == ' ':
        board[4] = 'O'
        return 4
    
    # Rule 4: Take first empty
    for i in range(9):
        if board[i] == ' ':
            board[i] = 'O'
            return i

# Game Simulation
moves = []
for turn in range(9):
    if turn % 2 == 0:
        # Player move (auto for demo)
        for i in range(9):
            if board[i] == ' ':
                board[i] = 'X'
                moves.append(f"X->{i}")
                break
    else:
        move = ai_move()
        moves.append(f"O->{move}")
    
    if check_win('X') or check_win('O'):
        break

# Output
arrow_path = " -> ".join(moves)
print("\nMoves List:", moves)
print("Moves Path:", arrow_path)

# DataFrame
print("\n", pd.DataFrame(moves, columns=["Moves"]))

# Final Board
print("\nFinal Board:")
print_board()

# Visualization (simple grid)
plt.figure()
plt.title("Tic-Tac-Toe Board")

for i in range(9):
    x, y = i % 3, 2 - i // 3
    plt.text(x, y, board[i], fontsize=20, ha='center')

plt.xlim(-0.5,2.5)
plt.ylim(-0.5,2.5)
plt.grid()
plt.show()
```
