```python
import matplotlib.pyplot as plt
import networkx as nx

# Graph & Heuristic
graph = {
    'A': ['B','C'],
    'B': ['D','E'],
    'C': ['F'],
    'D': [], 'E': [], 'F': []
}
h = {'A':1,'B':3,'C':4,'D':5,'E':6,'F':2}

# Hill Climbing
def hill_climb(g, h, start):
    path = [start]
    while g[start]:
        next_node = max(g[start], key=lambda x: h[x])
        if h[next_node] <= h[start]: break
        path.append(next_node)
        start = next_node
    return path

# Run
path = hill_climb(graph, h, 'A')
print("Hill Climbing Path:", " -> ".join(path))

# Visualization
G = nx.DiGraph([(u, v) for u in graph for v in graph[u]])
pos = nx.spring_layout(G)

nx.draw(G, pos, with_labels=True, node_size=2000)
nx.draw_networkx_edges(G, pos,
    edgelist=list(zip(path, path[1:])), width=2)

plt.title("Hill Climbing")
plt.show()
```
